```python
# Outliers
Q1 = data['price'].quantile(0.25)
Q3 = data['price'].quantile(0.75)

IQR = Q3 - Q1

print(Q1,Q3,IQR)
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = data[(data['price'] < lower_limit) |
                (data['price'] > upper_limit)]

print(outliers)

data_clean = data[(data['price'] >= lower_limit) &
                  (data['price'] <= upper_limit)]

print("Original Shape:", data.shape)
print("New Shape:", data_clean.shape)
```

```python
x = data.drop("price", axis=1)  # drop target from features
y = data["price"]               # target column
```

```python
x
```

```python
scale=StandardScaler()
scale.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.score(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
predicted_price = model.predict([[14000, 2, 2, 2, 1, 0, 1, 0, 1]])
print("Predicted House Price: ₹", round(predicted_price[0], 2))
```
